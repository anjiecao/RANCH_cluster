# Generate Param info 

```python
params = {

    # constant / not searching 
    'linking_hypothesis': ['EIG'], # possible values: 'EIG', 'KL', 'surprisal', 'quad_surprisal','no_noise', "no_learning"
    'mu_prior': [0],
    'hypothetical_obs_grid_n': [5], 
    'max_observation': [500],

    # only varies for quad_surprisal: 
    'quad_a': [999], 
    'quad_b': [999], 
    'quad_c': [999],

    # searching 
    'V_prior': np.arange(1, 4, 1).tolist(),
    'alpha_prior': [0.1,1,10],
    'beta_prior': [0.1,1,10],
    'epsilon': [0.0001],
    'mu_epsilon': [0.001],
    'sd_epsilon': [0.1,0.5,1], 
    'world_EIGs': [0.0001, 0.00001], 
    'forced_exposure_max':[5], # for KL try ~10e-5, for surprisal between 1 and 10

}

all_params_combinations = list(itertools.product(*list(params.values())))


params_info = pd.DataFrame(list(all_params_combinations), columns=list(params))
params_info = params_info.reset_index()
params_info = params_info.rename(columns={'index': 'param_id'})

params_info.to_csv("sim_info/param_info/eig.csv")
```
